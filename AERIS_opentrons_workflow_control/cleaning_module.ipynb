# Dry Pads

```python
# Load Labware
drypads = protocol.load_labware("fgl_12_tall_drypad_rack", 4)
```

```python
def clean(plate, sponge_loc: int, num_of_repetitions: int):
    """Clean a plate using a specified sponge location and number of repetitions. 
    To clean with multiple sponges, call this function multiple times with different sponge_loc values. 
    
    Parameters:
    plate: CaberPlate object - the plate to be cleaned (e.g. "sixmm")
    sponge_loc: the location of the sponge to be used
    num_of_repetitions: the number of times to repeat the cleaning process
    
    """
    
    right.pick_up_tip(drypads.wells()[sponge_loc])
    right.move_to(plate.get_pltloc().move(Point(z = 4.5)))
    
    for rep in range(num_of_repetitions):
        right.move_to(plate.get_pltloc().move(Point(z = 4.5, y = 1)), speed = 10)
        right.move_to(plate.get_pltloc().move(Point(z = 4.5, y = -1)), speed = 10)
    
    right.return_tip()
```
